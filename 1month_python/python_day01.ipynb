```python
# pow 함수

print(3 ** 5)
print(pow(3, 5)) # 3의 5제곱
```

```python
# divmod 함수

a = 4096
b = 7
몫 = a // b
나머지 = a % b

print(몫, 나머지)

몫, 나머지 = divmod(a, b) # 몫과 나머지를 튜플로 반환
print(몫, 나머지)
```

```python
# math 내장라이브러리

import math

pi = math.pi #원주율

print(math.sqrt(37)) # 제곱근
print(math.isqrt(37)) # 제곱근의 정수부분
print(math.ceil(37.5)) # ceiling function (주어진 수보다 작지 않은 최소의 정수)
print(math.floor(37.5)) # floor function (주어진 수보다 크지 않은 최대의 정수)
print(math.sin(pi * 3 / 4)) # sin 값
print(math.radians(60)) # 도를 라디안 값으로 변환
print(math.cos(math.radians(75))) # cos 75도
```
