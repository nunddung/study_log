```python
# 문자열 포맷팅
# C-style Formatting
# list, tuple, dict 등의 변수 표시할 때 오류가 자주 발생함

print("integer: %d, float: %.3f" % (3, 3.11592))
print("pi: %.6f" % (3.141512432449))
```

```python
# curly bracket Formatting

print("I am {} and pi value is {}".format('developer', 3.1415926535))
print("I am {} and pi value is {:.6f}".format('developer', 3.1415926535))

# 순서대로 값을 넣는 것 뿐만아니라 placeholder에 이름을 붙이고 그 위치마다 값을 넣어줄 수 있음

print("{이름}님의 잔고: 1억원, {이름}님의 은행은 {은행이름}입니다. {은행이름}의 서비스를 이용해주셔서 감사합니다. {이름} 고객님".format(이름='남예준', 은행이름='플레이브은행'))
print("pi value = {pi:.3f}".format(pi=3.141592)) # 이름이 정해진 상태에서 소수점 자르기도 가능

students = ['suzan', 'kevin', 'bob']
print("Students are {}".format(students)) # list도 출력가능
```

```python
# f-string
# 직접 변수를 문자열에 포맷팅한 문자열을 만들 수 있음
name = '박재형'
print(f"안녕하세요 {name} 고객님")

pi = 3.14592653589
print(f"pi = {pi:.3f}")

new_tuple = 1, 2, 3, 4
score = {"kevin" : 10, "suzan" : 3}
print(f"{new_tuple =}")
print(f"{score = }")
```

```python
# Boolean Type

print(1 == 1)
print(1 == 2)

# if문(조건문)

condition = False

if condition: # if가 True이면 출력한다
  print("Hello, True World")
else:
  print("Hello, False World")

score = 75
if score >= 90:
  print("Grade is A")
elif score >= 80:
  print("Grade is B")
else:
  print("Grade is C")
```

```python
# for문(반복문)
# iterable문 = list, dic, tuple, 문자열 등
for i in [1, 1, 2, 3, 5, 7, 12, 19]: # list와 같은 iterable 객체요소로 변수를 할당하고 이에 따라 동작을 반복
  print(i+5)
```

```python
# while문
a = 1
while a < 10:
  print(a)
  a += 2
```

```python
# while문으로 몫과 나머지 구하기

a = 65523
b =37
q = 0
r = 0

while a >= b:
  a -= b
  q += 1

r = a

print(f"{q=} {r=}")
```
